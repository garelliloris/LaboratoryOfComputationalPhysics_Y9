2\. Convert the following function into a pure function with no global variables or side effects

````python
x = 5
def f(alist):
    for i in range(x):
         alist.append(i)
    return alist

alist = [1,2,3]
ans = f(alist)
print (ans)
print (alist) # alist has been changed!
````

````python
def f(l):
    x = 5
    t=l.copy()
    for i in range(x):
         t.append(i)
    return t

alist = [1,2,3]
ans = f(alist)
print (ans)
print (alist) # alist has NOT been changed!
````

3\. Write a `decorator` hello that makes every wrapped function print “Hello!”, i.e. something like:

```python
@hello
def square(x):
    return x*x
```



````python
def hello(func):
    def wrapper(x):
        print("Hello!")
        y=func(x)
        print (y)
        print("Hello!")
        return y
    return wrapper

@hello
def square(x):
    return x*x

x=2
y=square(x)
print(y)
````

4\. Write the factorial function so that it a) does and b) does not use recursion.

````python
def factorial(n): 
    
    # Checking the number
    # is 1 or 0 then
    # return 1
    # other wise return
    # factorial
    if (n==1 or n==0):
        
        return 1
    
    else:
        
        return (n * factorial(n - 1)) 

num = 5; 
print("number : ",num)
print("Factorial : ",factorial(num))
#Recursion
%timeit factorial(20)
````

````python
def factorial(n):
    f=1
    for i in range (1,n+1):
        f=f*i
    return f

num = 5; 
print("number : ",num)
print("Factorial : ",factorial(num))
#NO recurision
%timeit factorial(20)
````

5\. Use HOFs (zip in particular) to compute the weight of a circle, a disk and a sphere, assuming different radii and different densities:

```python
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
radii = [1,2,3]
```

where the entries of the dictionary's values are the linear, superficial and volumetric densities of the materials respectively.

In particular define a list of three lambda functions using a comprehension that computes the circumference, the area and the volume for a given radius.

````python
import math
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
radii = [1,2,3]
weights={}
vol=[lambda r: 2*math.pi*r, lambda r: math.pi*r**2, lambda r: 4*math.pi*r**3/3]
for material,rho in densities.items():
    weights[material]=[
        [density*v(r) for r in radii] for density,v in zip(rho,vol)
    ]
    print (f"{material} with radius {radii} has linear weight {weights[material][0]} \n has superficial weight {weights[material][1]} \n has volumetric weight {weights[material][2]}")
````

````python
# Parent class
class Dog:

    # Class attribute
    species = 'mammal'

    # Initializer / Instance attributes
    def __init__(self, name, age, is_hungry):
        self.name = name
        self.age = age
        self.is_hungry = is_hungry

    # instance method
    def description(self):
        return "{} is {} years old".format(self.name, self.age)

    # instance method
    def speak(self, sound):
        return "{} says {}".format(self.name, sound)

    def eat(self):
        self.is_hungry = False
        return "{} is not hungry anymore".format(self.name, self.is_hungry)

# Child class (inherits from Dog class)
class RussellTerrier(Dog):
    def run(self, speed):
        return "{} runs {}".format(self.name, speed)

# Child class (inherits from Dog class)
class Bulldog(Dog):
    def run(self, speed):
        return "{} runs {}".format(self.name, speed)


Tom=Dog("Tom",6,True)
Fletcher=Dog("Fletcher",7,True)
Larry=Dog("Larry",9,True)
Dogs=[Tom,Fletcher,Larry]



print(f"I have {len(Dogs)} dogs.")
for Dog in Dogs:
    print (f"{Dog.name} is {Dog.age}.")
    Dog.eat()
if Dog.species == 'mammal':
    print(f"And they're all {Dog.species}s, of course.")
if all(Dog.is_hungry for Dog in Dogs):
    print("My dogs are hungry.")
elif all(Dog.is_hungry==False for Dog in Dogs):
    print("My dogs are not hungry.")
````

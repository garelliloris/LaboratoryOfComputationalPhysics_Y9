1\. Write a function that converts number representation, bin<->dec<->hex. (Clearly using the corresponding python built-in functions is not fair..)

```python
def to_decimal(n,bi):
    digits=list(map(int,n.split()))
    decimal=0
    for digit in digits:
        decimal=decimal*bi+digit
    return decimal

def from_decimal(n,bo):
    if n==0:
        return[0]
    else:
        digits = []
        while n:
            digits.append(int(n % bo))
            n //= bo
        return digits[::-1]

print('Enter the number to convert:(digits separated by spaces) ')
number=input()
print('Enter the base of the number in input: ')
base_in=int(input())
print('Enter the base you want to convert the number to: ')
base_out=int(input())
number_dec=to_decimal(number,base_in)
if base_out == 10:
    number_out=number_dec
else:
    number_out=from_decimal(number_dec,base_out)
print(f'The number {number} in base {base_out} is: \n {number_out}')
```

2\. Write a function that converts a 32 bit word into a single precision floating point (i.e. interprets the various bits as sign, mantissa and exponent)

```python
def to_decimal(n, bi):
    digits = list(map(int, n.split()))
    bi = int(bi)
    decimal = 0
    for digit in digits:
        decimal = decimal * bi + digit
    return decimal

def interpret(l):
    sign = l[0]
    exp = l[1:9]
    mant = l[9:]
    
    if sign == 0:
        s = 1
    else:
        s = -1
    exp_val = to_decimal(' '.join(map(str, exp)), 2)
    mant_val = to_decimal(' '.join(map(str, mant)), 2)

    if exp_val == 255:
        if mant_val == 0:
            # If mantissa is 0, is Inf (+ or -)
            return float('inf') if s == 1 else float('-inf')
        else:
            # If mantissa nor 0, is NaN (Not a Number)
            return float('nan')
    elif exp_val == 0:
        if mant_val == 0:
            # Zero (+0.0 o -0.0)
            return 0.0 if s == 1 else -0.0
        else:
            # Subnormal numbers
            m = mant_val / (2**23)
            return s * (2 ** -126) * m
    m=mant_val/(2**23)
    e=exp_val-127
    
    n = s * (2 ** e) * (1 + m)
    return n

print("Enter a 32 bit word (one bit per line):")
word = []

for i in range(32):
    while True:
        try:
            bit = int(input(f"Bit {i+1}: "))
            if bit in (0, 1):
                word.append(bit)
                break
            else:
                print("Error: you must insert 0 or 1.")
        except ValueError:
            print("Error: insert a valid number (0 or 1).")

print(f"Word correctly written (lenght: {len(word)})")
number = interpret(word)
print(number)
```

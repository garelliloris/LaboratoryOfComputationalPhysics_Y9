1\. Write a function that converts number representation, bin<->dec<->hex. (Clearly using the corresponding python built-in functions is not fair..)

```python
def to_decimal(n,bi):
    digits=list(map(int,n.split()))
    decimal=0
    for digit in digits:
        decimal=decimal*bi+digit
    return decimal

def from_decimal(n,bo):
    if n==0:
        return[0]
    else:
        digits = []
        while n:
            digits.append(int(n % bo))
            n //= bo
        return digits[::-1]

print('Enter the number to convert:(digits separated by spaces) ')
number=input()
print('Enter the base of the number in input: ')
base_in=int(input())
print('Enter the base you want to convert the number to: ')
base_out=int(input())
number_dec=to_decimal(number,base_in)
if base_out == 10:
    number_out=number_dec
else:
    number_out=from_decimal(number_dec,base_out)
print(f'The number {number} in base {base_out} is: \n {number_out}')
```

2\. Write a function that converts a 32 bit word into a single precision floating point (i.e. interprets the various bits as sign, mantissa and exponent)

```python
def to_decimal(n, bi):
    digits = list(map(int, n.split()))
    bi = int(bi)
    decimal = 0
    for digit in digits:
        decimal = decimal * bi + digit
    return decimal

def interpret(l):
    sign = l[0]
    exp = l[1:9]
    mant = l[9:]
    
    if sign == 0:
        s = 1
    else:
        s = -1
    exp_val = to_decimal(' '.join(map(str, exp)), 2)
    mant_val = to_decimal(' '.join(map(str, mant)), 2)

    if exp_val == 255:
        if mant_val == 0:
            # If mantissa is 0, is Inf (+ or -)
            return float('inf') if s == 1 else float('-inf')
        else:
            # If mantissa nor 0, is NaN (Not a Number)
            return float('nan')
    elif exp_val == 0:
        if mant_val == 0:
            # Zero (+0.0 o -0.0)
            return 0.0 if s == 1 else -0.0
        else:
            # Subnormal numbers
            m = mant_val / (2**23)
            return s * (2 ** -126) * m
    m=mant_val/(2**23)
    e=exp_val-127
    
    n = s * (2 ** e) * (1 + m)
    return n

print("Enter a 32 bit word (one bit per line):")
word = []

for i in range(32):
    while True:
        try:
            bit = int(input(f"Bit {i+1}: "))
            if bit in (0, 1):
                word.append(bit)
                break
            else:
                print("Error: you must insert 0 or 1.")
        except ValueError:
            print("Error: insert a valid number (0 or 1).")

print(f"Word correctly written (lenght: {len(word)})")
number = interpret(word)
print(number)
```

3\. Write a program to determine the underflow and overflow limits (within a factor of 2) for python on your computer. 

**Tips**: define two variables inizialized to 1 and halve/double them enough time to exceed the under/over-flow limits  

```python
x=1.0

while True:
    try:
        next_val=2.0*x
        
        if next_val==float('inf'):
            print("Inf reached")
            break
        else:
            x=next_val
    except:
        print('Overflow error')
        break

print(f"Last valid value before overflow: {x}")
y=1.0

while True:
    next_val=y/2

    if next_val==0.0:
        print('Zero reached')
        break
    else:
        y=next_val

print (f'Last valid value before overflow: {y}')
```

```python
def std_root(a,b,c):
    x1=(-b+(b**2-4*a*c)**0.5)/(2*a)
    x2=(-b-(b**2-4*a*c)**0.5)/(2*a)
    return x1,x2
def redef_root(a,b,c):
    x1=2*c/(-b+(b**2-4*a*c)**0.5)
    x2=2*c/(-b-(b**2-4*a*c)**0.5)
    return x1,x2
def exact_root(a,b,c):
    if b>=0:
        x1=2*c/(-b-(b**2-4*a*c)**0.5)
        x2=(-b-(b**2-4*a*c)**0.5)/(2*a)
    else:
        x1=(-b+(b**2-4*a*c)**0.5)/(2*a)
        x2=2*c/(-b+(b**2-4*a*c)**0.5)
    return x1,x2
a=0.001 
b=-1000 
c=0.001
x1,x2=std_root(a,b,c)
print (f'a:{x1,x2}')
x1,x2=redef_root(a,b,c)
print (f'b:{x1,x2}')
x1,x2=exact_root(a,b,c)
print(f'c:{x1,x2}')
```

7\. Consider the integral of the semicircle of radius 1:
$$
I=\int_{-1}^{1} \sqrt(1-x^2) {\rm d}x
$$
which it's known to be $I=\frac{\pi}{2}=1.57079632679...$.
Alternatively we can use the Riemann definition of the integral:
$$
I=\lim_{N\to\infty} \sum_{k=1}^{N} h y_k 
$$

with $h=2/N$ the width of each of the $N$ slices the domain is divided into and where
$y_k$ is the value of the function at the $k-$th slice.

(a) Write a programe to compute the integral with $N=100$. How does the result compares to the true value?

(b) How much can $N$ be increased if the computation needs to be run in less than a second? What is the gain in running it for 1 minute?
